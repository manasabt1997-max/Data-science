```python
#1. install required tools
#install cutadapt
!pip install cutadapt
#install fastQC for QC checks
!apt-get install fastqc -y
#update package list
!apt-get update -q
#install sra toolkit directly from ubuntu repository
!apt-get install -y sra-toolkit
#verify installation
!fasterq-dump --version
```

```python
#2. download the data
#download the data
!prefetch SRR957824
#convert to fastq file
!fasterq-dump SRR957824 --progress
#upload session File
!ls
```

```python
#3. initial fastqc
!fastqc SRR957824_1.fastq
!fastqc SRR957824_2.fastq
```

```python
#5. adapter trimming with cutadapt- removes universal adapter and overrepresented sequence
!cutadapt \
  -a AGATCGGAAGAGCACACGTCTGAACTCCAG \
  -A GACTACHVGGGTATCTA \
  -o trimmed-cutadapt-1.fastq \
  -p trimmed-cutadapt-2.fastq \
  --minimum-length 36 \
  -q 15,15 \
  SRR957824_1.fastq SRR957824_2.fastq > cutadapt-log.txt
```

```python
#6. post trim fastqc- run fastqc on oth trimmed outputs
!fastqc trimmed-cutadapt-1.fastq
!fastqc trimmed-cutadapt-2.fastq
```

```python
#7. indexing the reference genome
#install BWA
!apt-get install -y bwa

# Create reference directory if it doesn't exist
!mkdir -p /content/reference

# Download a genuine, complete E. coli reference genome (K-12 MG1655) from NCBI
!wget -O /content/reference/SRR957824.fa.gz "https://ftp.ncbi.nlm.nih.gov/genomes/all/GCF/000/005/845/GCF_000005845.2_ASM584v2/GCF_000005845.2_ASM584v2_genomic.fna.gz"
# Gunzip to get the plain FASTA file
!gunzip -f /content/reference/SRR957824.fa.gz

# Index the valid reference genome using bwa
!bwa index /content/reference/SRR957824.fa
```

```python
#8. fetch 1000 reads from the test datasets(SRR957824)
!fastq-dump -X 1000 --fasta SRR957824 -O .
```

```python
!bwa mem -M -R '@RG\tID:sample\tPL:Illumina' /content/reference/SRR957824.fa trimmed-cutadapt-1.fastq trimmed-cutadapt-2.fastq > output.sam
```

```python
#9. install sam tools
!apt-get update -y
!apt-get install -y samtools
#verify the installation
!samtools --version
```

```python
#inspect top of the file
!head output.sam
```

```python
#11. convert sam to bam
!samtools view -b output.sam -o raw_unsorted.bam
```

```python
!ls -1h output.sam raw_unsorted.bam
```

```python
#sort our bam alignment
!samtools sort -o sorted_alignments.bam raw_unsorted.bam
```

```python
#lets check the file existence
!ls sorted_alignments.bam
```

```python
#12. Index the bam file
!samtools index sorted_alignments.bam
```

```python
#verify its existence in the file (.bai)
!ls -- -sorted_alignments.bam
```

```python
#filter out unmapped reads
!samtools view -b -f 4 sorted_alignments.bam > filtered_alignments.bam
```

```python
#count the reads to verify our filteration worked successfully
!samtools view -c sorted_alignments.bam
print ('\n total recoreds remaining after filtering out unmapped reads:')
!samtools view -c final_clean_mapped.bam
!samtools view sorted_alignments.bam
```
