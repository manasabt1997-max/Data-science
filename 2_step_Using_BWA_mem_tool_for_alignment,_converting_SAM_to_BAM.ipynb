```python
#download human hg19chr1 reference
!wget -q https://hgdownload.soe.ucsc.edu/goldenpath/hg38/chromosomes/chr19.fa.gz
!gunzip chr19.fa.gz
```

```python
#index the reference
!bwa index chr19.fa
```

```python
#fetch 1000 reads from a test dataset (SRR957824)
! fastq-dump -X 1000 SRR957824 >reads-hg19.fq
```

```python
#align reads (output to aligned_hg19.sam)
!bwa mem chr19.fa reads-hg19.fq > aligned_hg19.sam
```
