PCA & T-SNE

```python
import pandas as pd
from sklearn.manifold import TSNE
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
data = {'MW':[180,250,300,350,500,600,150,220,400,480],
        'logP':[2.5,3.2,4.0,4.5,5.2,6.0,1.2,2.8,4.8,5.5],
        'TPSA':[75,80,65,60,55,40,95,85,50,45],
        'HBD':[2,1,1,0,0,0,3,2,0,0],
        'HBA':[5,6,6,7,8,8,4,5,7,8]}
df = pd.DataFrame(data)
```

```python
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
```

```python
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_scaled)
```

```python
plt.scatter(pca_result[:, 0], pca_result[:, 1], c='blue')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA: Module clusters')
plt.show()
```

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=3)
tsne_result = tsne.fit_transform(df_scaled)
plt.scatter(tsne_result[:, 0], tsne_result[:, 1], c='blue')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.title('t-SNE: Module clusters')
```
