```python
#import librarirs
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder # #to transform dependent catagorical features to numerical values
from sklearn.preprocessing import OrdinalEncoder #to transform independent catagorical features to numerical values
from matplotlib import pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix
from sklearn.impute import KNNImputer
```

```python
#load the dataset
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
```

```python
from sklearn import feature_selection
#labeling the data and feature selection
#ML models require features(inputs)=(X) and targets(output)=(Y) as separete variables
#data.data= the raw independent variable[contains numbers like tumour size and radius]
#data.target= the dependent variables (labels to predict)[contains 0=malignant, 1= benign]
x = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.DataFrame(data.target, columns=['target'])
x.head()
x.isnull().sum()
```

```python
imputer = KNNImputer(n_neighbors=5)
x = imputer.fit_transform(x)
```

```python
#train and test the model
#split 80% training data and 20% testing data
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = RandomForestClassifier(n_estimators=100)# n_estimators=100 is used bcz its a safe number to get the model accuracy if incresed accuracy increases but takes time
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
roc_auc_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(model, x_test, y_test, cmap='Blues')
plt.show()
```

According to the prediction we can conclude its a good model

```python
#data cleaning and normalization
#KNN imputation
from sklearn.impute import KNNImputer
from sklearn.preprocessing import LabelEncoder

# Drop 'id' and 'Unnamed: 32' columns as they are not suitable for imputation
# and 'Unnamed: 32' is entirely null.
df_cleaned = df.drop(columns=['id', 'Unnamed: 32'], errors='ignore')

# Encode 'diagnosis' column as KNNImputer expects numerical input
le = LabelEncoder()
df_cleaned['diagnosis'] = le.fit_transform(df_cleaned['diagnosis'])

imputer = KNNImputer(n_neighbors=5)

# Apply imputation and convert the resulting NumPy array back to a DataFrame
df_imputed_array = imputer.fit_transform(df_cleaned)
df = pd.DataFrame(df_imputed_array, columns=df_cleaned.columns)
```

```python
#feature selection
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel # Import SelectFromModel

Model = RandomForestClassifier(n_estimators=100)
# Fit the model on the full dataset before using it for feature selection
# y needs to be flattened from a DataFrame to a 1D array
Model.fit(x, y.values.ravel())

# Corrected typo: threshodl to threshold, and using Model as estimator
feature_selector = SelectFromModel(estimator=Model, threshold='median')

#fit and transform
# y needs to be flattened from a DataFrame to a 1D array
x_new = feature_selector.fit_transform(x, y.values.ravel())
#get selected names
selected_features_indices = feature_selector.get_support(indices=True)
# Use data.feature_names as x is a numpy array and no longer has 'columns'
selected_features_names = data.feature_names[selected_features_indices]
print(f'features selected from RandomForest: {selected_features_names}')
```

```python
#Model training
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x_new, y, test_size= 0.3, random_state=42)

#initialize and fit the model
model= RandomForestClassifier(n_estimators=100, random_state= 42)
model.fit(x_train, y_train)
```

```python
#generate predictions on the test partition
y_pred = model.predict(x_test)
y_pred_proba = model.predict_proba(x_test)[:, 1]
```

```python
#Evaluation metrics
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, classification_report
print('classification summary:\n')
print(classification_report(y_test, y_pred))
print(f'accuracy: {accuracy_score(y_test, y_pred)}')
print(confusion_matrix(y_test, y_pred))
```
