```python
!pip install biopython
from Bio.Seq import Seq
from Bio import SeqIO
```

```python
#create a DNA sequence
dna_seq = Seq("ATCGATCGATCG")
print ("DNA sequence:", dna_seq)
```

```python
#transcription
rna_seq = dna_seq.transcribe()
print ("RNA sequence:", rna_seq)
```

```python
#translate
protein_seq = dna_seq.translate()
print ("Protein sequence:", protein_seq)
```

```python
#2. write a FASTA file
fasta_content = ">example1\nATGGCCATTGG"
file_path = "/content/sequence.fasta"
with open(file_path, "w") as f:
    f.write(fasta_content)
    #read using SeqIO
for record in SeqIO.parse('/content/sequence.fasta', "fasta"):
    print("ID:", record.id)
    print("Sequence:", record.seq)
```

```python
import gzip
from Bio.PDB import MMCIFParser

# Path to your compressed file
gzipped_file_path = "/content/2C0K (1).cif.gz"
cif_file_path = "/content/2C0K.cif"

# Decompress the file
with gzip.open(gzipped_file_path, "rb") as f_in:
    with open(cif_file_path, "wb") as f_out:
        f_out.write(f_in.read())

# Parse using MMCIFParser
mmcif_parser = MMCIFParser()
structure = mmcif_parser.get_structure("hemoglobin", cif_file_path)
print("Structure loaded successfully:", structure)
```
